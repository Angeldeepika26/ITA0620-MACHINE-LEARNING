```python
import csv

data = [
    ['High', 'Yes', 'No', 'Yes'],
    ['High', 'Yes', 'Yes', 'No'],
    ['Low', 'Yes', 'No', 'No'],
    ['High', 'No', 'No', 'No'],
    ['High', 'Yes', 'No', 'Yes'],
    ['Low', 'No', 'Yes', 'No'],
    ['High', 'Yes', 'No', 'Yes'],
    ['Low', 'Yes', 'No', 'No']
]

h = ['0', '0', '0']

for row in data:
    if row[3] == 'Yes':
        x = row[:3]
        for i in range(3):
            if h[i] == '0':
                h[i] = x[i]
            elif h[i] != x[i]:
                h[i] = '?'

print("Final Hypothesis:", h)
```
