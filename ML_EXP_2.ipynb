```python
data = [
    ['High', 'Yes', 'No', 'Yes'],
    ['High', 'Yes', 'Yes', 'No'],
    ['Low', 'Yes', 'No', 'No'],
    ['High', 'No', 'No', 'No'],
    ['High', 'Yes', 'No', 'Yes'],
    ['Low', 'No', 'Yes', 'No'],
    ['High', 'Yes', 'No', 'Yes'],
    ['Low', 'Yes', 'No', 'No']
]

S = ['0', '0', '0']
G = [['?', '?', '?']]

for row in data:
    x, target = row[:3], row[3]

    if target == 'Yes':
        for i in range(3):
            if S[i] == '0':
                S[i] = x[i]
            elif S[i] != x[i]:
                S[i] = '?'

        G = [g for g in G
             if all(g[i] == '?' or g[i] == x[i]
                    for i in range(3))]

    else:
        new_G = []
        for g in G:
            if all(g[i] == '?' or g[i] == x[i]
                   for i in range(3)):
                for i in range(3):
                    if g[i] == '?' and S[i] != '?':
                        h = g.copy()
                        h[i] = S[i]
                        new_G.append(h)
            else:
                new_G.append(g)

        G = new_G

print("Specific Boundary (S):", S)
print("General Boundary (G):", G)
```
