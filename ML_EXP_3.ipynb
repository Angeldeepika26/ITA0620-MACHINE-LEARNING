```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier, export_text

data = {
    'TransactionAmt': ['High', 'High', 'Low', 'High',
                       'High', 'Low', 'High', 'Low'],
    'ForeignIP': ['Yes', 'Yes', 'Yes', 'No',
                  'Yes', 'No', 'Yes', 'Yes'],
    'CardPresent': ['No', 'Yes', 'No', 'No',
                    'No', 'Yes', 'No', 'No'],
    'Fraud': ['Yes', 'No', 'No', 'No',
              'Yes', 'No', 'Yes', 'No']
}

df = pd.DataFrame(data)

X = pd.get_dummies(df.drop('Fraud', axis=1))
y = df['Fraud']

model = DecisionTreeClassifier(
    criterion='entropy',
    random_state=0
)

model.fit(X, y)

print("Decision Tree:")
print(export_text(model, feature_names=list(X.columns)))

sample = pd.DataFrame([{
    'TransactionAmt': 'High',
    'ForeignIP': 'Yes',
    'CardPresent': 'No'
}])

sample = pd.get_dummies(sample).reindex(
    columns=X.columns, fill_value=0
)

prediction = model.predict(sample)
print("Predicted Fraud:", prediction[0])
```

```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier, export_text

data = [
    ["Sunny",    "Hot",  "High",   "Weak",   "-"],
    ["Sunny",    "Hot",  "High",   "Strong", "-"],
    ["Overcast", "Hot",  "High",   "Weak",   "+"],
    ["Rain",     "Mild", "High",   "Weak",   "+"],
    ["Rain",     "Cool", "Normal", "Weak",   "+"],
    ["Rain",     "Cool", "Normal", "Strong", "-"],
    ["Overcast", "Cool", "Normal", "Strong", "+"],
    ["Sunny",    "Mild", "High",   "Weak",   "-"],
    ["Sunny",    "Cool", "Normal", "Weak",   "+"],
    ["Rain",     "Mild", "Normal", "Weak",   "+"],
    ["Sunny",    "Mild", "Normal", "Strong", "+"],
    ["Overcast", "Mild", "High",   "Strong", "+"],
    ["Overcast", "Hot",  "Normal", "Weak",   "+"],
    ["Rain",     "Mild", "High",   "Strong", "-"]
]

# Create DataFrame with correct column names
df = pd.DataFrame(
    data,
    columns=["Outlook", "Temperature", "Humidity", "Wind", "Play"]
)

# Separate input features and target
X = pd.get_dummies(df.drop("Play", axis=1))
y = df["Play"]

# Train Decision Tree
model = DecisionTreeClassifier(
    criterion="entropy",
    random_state=0
)

model.fit(X, y)

# Display Decision Tree
print("Decision Tree:")
print(export_text(model, feature_names=list(X.columns)))

# Predict for a new sample
sample = pd.DataFrame([{
    "Outlook": "Sunny",
    "Temperature": "Cool",
    "Humidity": "Normal",
    "Wind": "Strong"
}])

sample = pd.get_dummies(sample).reindex(
    columns=X.columns, fill_value=0
)

prediction = model.predict(sample)

print("Predicted Play:", prediction[0])
```
